##7.1.1 Hukum Ohm

Hukum Ohm menyatakan hubungan antara tegangan, arus listrik dan hambatan dalam suatu rangkaian listrik. Pada hukum Ohm diperlukan gaya untuk mendorong suatu muatan.
  
Dalam membuat arus, agar arus mengalir Anda harus mendorong muatan-muatan tersebut. Seberapa cepat mereka bergerak, sebagai respons terhadap dorongan yang diberikan, tergantung pada sifat materialnya. Untuk sebagian besar zat, rapat arus **J** sebanding dengan gaya per satuan muatan, **f**:                
$${\bf J}=\sigma \, {\bf f} \qquad \qquad         ....... \qquad               (7.1)$$

Faktor proporsionalitas $\sigma$ adalah konstanta empiris yang bervariasi dari satu bahan ke bahan lainnya itu disebut **konduktivitas** dari medium. Sedangkan itu kebalikan dari $\sigma$ , disebut yang **resistivitas**: $\rho = \frac{1}{\sigma}$. Beberapa tipikal nilai-nilai tercantum dalam Tabel 7.1. Perhatikan bahwa bahkan isolator bekerja sedikit meskipun konduktivitas logam secara astronomis lebih besar. Sebenarnya, untuk sebagian besar tujuan
logam dapat dianggap sebagai konduktor yang sempurna, dengan $\sigma = \infty$,  sedangkan untuk isolator
kita bisa menganggap $\sigma=0$.

Pada prinsipnya, gaya yang mendorong muatan untuk menghasilkan arus bisa berupa apa saja - kimiawi, gravitasi,  gaya elektromagnetik yang melakukan pekerjaan itu. Gaya yang mendorong muatan ini merupakan gaya elektromagnet. Dalam hal ini Persamaan 7.1 menjadi:

$${\bf J}=\sigma({\bf E}+{v}\times{\bf B}) \qquad \qquad .......... \qquad (7.2)$$

Biasanya, kecepatan muatan cukup kecil sehingga suku kedua dapat diabaikan:

$${\bf J}=\sigma{\bf E}\qquad \qquad \qquad \qquad .......... \qquad (7.3)$$

Persamaan 7.3 disebut **Hukum Ohm**. Selain itu, untuk konduktor sempurna ${\bf E}=\frac{\bf J}{\sigma}=0$ meskipun arus mengalir. Dalam praktiknya, logam adalah konduktor yang sangat baik sehingga medan listrik yang dibutuhkan untuk menggerakkan arus di dalamnya dapat diabaikan. Jadi kami secara rutin memperlakukan kabel penghubung di sirkuit listrik (misalnya) sebagai ekuipotensial. **Resistor**, sebaliknya, terbuat dari bahan konduktor yang buruk.




###Example 7.1



Sebuah resistor silinder dengan luas penampang $A$ dan panjang $L$ dibuat dari bahan dengan konduktivitas $\sigma$ (Lihat Gambar 7.1) Penampang tidak harus melingkar, tapi itu adalah sama ke bawah. Jika kita menetapkan bahwa potensialnya konstan di setiap ujung, dan perbedaan potensial antara ujung-ujungnya adalah $V$, berapa arus yang mengalir?

<figcaption align="center"><b>Gambar 7.1</b> </figcaption>

**Solusi:**

Ternyata, medan listrik di dalam kawat seragam (saya akan membuktikannya di a saat). Ini mengikuti dari Persamaan (7.3) bahwa rapat arus juga seragam, jadi

$$I= JA$$

dimana, $J=\sigma E$ (Hukum Ohm), maka:

$$I=\sigma EA$$

dan $E = \frac {V} {L}$ (hukum beda potensial dengan medan listrik), sehingga didapatkan:

$$I=\frac{\sigma A}{L}V$$

```
sigma=0.5 #Konduktivitas
A=4 #Luas Penampang
V=200 #Beda Potensial
L=3 #Panjang
I=sigma*A*V/L
print(I)
```

###Example 7.2

Dua silinder logam koaksial panjang (jari-jari a dan b) dipisahkan
dengan bahan konduktivitas $\sigma$ (Lihat gambar 7.2). Jika dipertahankan pada suatu beda potensial $V$, arus apa yang mengalir dari satu ke yang lain, dalam panjang $L$?

<figure align="center"> <img src="https://drive.google.com/uc?id=1rQ0tbL-dMfibggVBKBsM64CbJ8xKWKCz"> </figure>
<figcaption align="center"><b>Gambar 7.2</b> </figcaption>

**Solusi:**

Medan antar silinder adalah

$$\vec E=\frac{\lambda}{2\pi\epsilon_{0} l} \hat l$$

dimana $\lambda$ adalah muatan per satuan panjang pada silinder dalam. Oleh karena itu saat ini

$$I=\int \vec J \cdot d {\vec a}$$

dari persamaan (7.3) bahwa $\vec J=\sigma \vec E$, maka

$$I=\int \sigma \vec E \cdot d \vec a$$

dimana $\sigma$ merupakan parameter, maka

$$I=\sigma\int \vec E \cdot d \vec a$$

Berdasarkan Hukum Gauss, medan yang merupakan salah satu dari empat Persamaan Maxwell adalah

$$\int \vec E \cdot d \vec a = \frac {Q} {\epsilon_{0}}$$

Integral dari medan yang menembus suatu permukaan sebanding dengan muatan listrik pada bidang itu.

$Q$ adalah muatan. Jadi, $Q$ = rapat muatan per satuan panjang dikali panjang atau $Q=\lambda L$.

$$I=\frac{\sigma}{\epsilon_{0}}\lambda L$$

(Integral terletak di atas semua permukaan yang menutupi silinder bagian dalam). Sementara itu, beda potensial antara silinder adalah

$$V=-\int\limits_b^a \vec E \cdot d \vec l $$

$$V=-\int\limits_b^a \frac{\lambda}{2\pi\epsilon_{0} l} \hat l \cdot d \vec l$$

dimana $d \vec l = dl \hat l$, maka

$$V=-\int\limits_b^a \frac{\lambda}{2\pi\epsilon_{0} l} \hat l \cdot dl \hat l$$

$$V=-\int\limits_b^a \frac{\lambda}{2\pi\epsilon_{0} l} dl$$

$$V=-\frac{\lambda}{2\pi\epsilon_{0}} \int\limits_b^a \frac{1}{l} dl$$

$$V=-\frac{\lambda}{2\pi\epsilon_{0}} \bigg[\ln l\bigg]_b^a$$

$$V=-\bigg(\frac{\lambda}{2\pi\epsilon_{0}}\bigg) (\ln a - \ln b)$$

$$V=\frac{\lambda}{2\pi\epsilon_{0}} (\ln b - \ln a)$$

$$V= \frac{\lambda}{2\pi\epsilon_{0}}\ln (\frac{b}{a})$$
jadi:
$$I=\frac{2\pi\sigma L}{ln (b/a)}V$$
